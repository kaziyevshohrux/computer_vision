```python
import matplotlib.pyplot as plt
import numpy as np

print(matplotlib.__version__)
```

```python
x = np.array([1,2,3,4,5,6])
y = np.array([10,20,25,32,40,10])

plt.plot(x, y)
plt.title("Basiv Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
from matplotlib.lines import lineStyles
x = np.array([1,2,3,4])
y = np.array([3,8,1,4])

plt.plot(y, linestyle="dotted", color="blue", marker="x" )
plt.title("styled line")
plt.show()
```

```python
x = np.array([1,2,3,4])
y1 = np.array([2,4,5,6])
y2 = np.array([1,3,3,8])

plt.plot(x, y1, label="Line 2", color="blue", marker="s")
plt.plot(x, y2, label="Line 2", color="red", marker="o" , linestyle="--")
plt.title("multiple lines")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.legend()
plt.show()
```

Bar Chart Numpy amd Maplotip

```python
x = np.array(["A", "B", "C", "D"])
y= np.array([5,12,25,4])

plt.bar(x, y, color="orange")
plt.title("Bar chart example")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

Numpy and Scatter Plot

```python
x = np.array([1,2,3,4,5,6])
y= np.array([10,45,65,123, -5, -12])

plt.scatter(x,y, color="pink", s=250)
plt.title("Scatter Plot example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

HIstogram with Numpy and Motplotib

```python
data = np.array([2,3,4,5,6,7,8,12])

plt.hist(data, bins=6, color="skyblue", edgecolor="black")
plt.xlabel("values")
plt.ylabel("counts")
plt.show()
```

Subplots(multiple graphs togeher)

```python
x = np.array([1,2,3,4])
y1 = np.array([2,4,6,8])
y2 = np.array([3,6,9,12])

plt.figure(figsize=(8,4))  # w,h

plt.subplot(1,2,1)
plt.plot(x, y1, color="blue")
plt.title("Plot 1")

plt.subplot(1,2,2)
plt.plot(x, y2, color="red")
plt.title("Plot 2")

plt.tight_layout()
plt.show()
```
